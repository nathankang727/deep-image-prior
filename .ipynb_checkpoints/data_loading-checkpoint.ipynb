```python
import numpy as np
import matplotlib.pyplot as plt

def preprocess_image(path, selected_channels=(0, 5, 7)):
    # Load and clip negative values
    data = np.load(path)
    data[data < 0] = 0

    # Select 3 specific channels (e.g., layers 0, 5, 7)
    selected_data = data[:, :, selected_channels]  # shape: (3, H, W)

    # Normalize each channel to [0, 255]
    normalized = np.zeros_like(selected_data, dtype=np.float32)
    for i in range(3):
        channel = selected_data[i]
        min_val = channel.min()
        max_val = channel.max()
        if max_val > min_val:
            normalized[i] = 255 * (channel - min_val) / (max_val - min_val)
        else:
            normalized[i] = 0

    # Rearrange axes to (H, W, 3) and convert to uint8
#     image = np.moveaxis(normalized, 0, -1).astype(np.uint8)
    image = normalized.astype(np.uint8)
    print(image)

    return image

# 100m image
image_100m = preprocess_image("/scratch/network/For_Nathan/input_data_100m.npy")
plt.imshow(image_100m)
plt.title("100m (RGB from layers 0, 5, 7)")
plt.axis("off")
plt.show()

# 200m image
image_200m = preprocess_image("/scratch/network/For_Nathan/input_data_200m.npy")
plt.imshow(image_200m)
plt.title("200m (RGB from layers 0, 5, 7)")
plt.axis("off")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def preprocess_image(path, size, selected_channels=(0, 5, 7), pressure_norm_mode="default"):
    # Load and clip negative values
    data = np.load(path)  # shape: (H, W, C)
    data[data < 0] = 0
    
    # Crop based on resolution
    if size == 100:
        imin, imax = 200, 420
        jmin, jmax = 100, 420
    else:
        imin, imax = 100, 210
        jmin, jmax = 50, 210

    selected_data = data[jmin:jmax, imin:imax, selected_channels]
    normalized = np.zeros_like(selected_data, dtype=np.float32)

    for i in range(3):
        channel = selected_data[:, :, i]

        if i == 0:  # Pressure channel
            channel = 1 - np.log(channel + 0.000001)
            if pressure_norm_mode == "default":
                min_val, max_val = channel.min(), channel.max()
            elif pressure_norm_mode == "percentile":
                min_val, max_val = np.percentile(channel, 2), np.percentile(channel, 98)
            elif pressure_norm_mode == "zscore":
                mean, std = channel.mean(), channel.std()
                channel = np.clip((channel - mean) / std, -2, 2)
                min_val, max_val = -2, 2
            elif pressure_norm_mode == "log":
                channel = np.log1p(channel)  # log(1 + x)
                min_val, max_val = channel.min(), channel.max()
            else:
                raise ValueError("Unknown pressure_norm_mode")

        else:  # Other channels
            min_val, max_val = channel.min(), channel.max()

        # Avoid divide-by-zero
        if max_val > min_val:
            normalized[:, :, i] = 255 * (channel - min_val) / (max_val - min_val)
        print(str(min_val) + " " + str(max_val))

    return normalized.astype(np.uint8)


modes = ["default", "percentile", "zscore", "log"]

for mode in modes:
    image = preprocess_image("/scratch/network/For_Nathan/input_data_100m.npy", 100, pressure_norm_mode=mode)
    plt.imshow(image, origin="lower")
    plt.title(f"100m (Pressure: {mode})")
    plt.axis("off")
    plt.show()
```

```python
def preprocess_image(path, size, selected_channels=(0, 5, 7), pressure_norm_mode="default", return_params=False):
    data = np.load(path)  # shape: (H, W, C)
    data[data < 0] = 0

    if size == 100:
        imin, imax = 200, 420
        jmin, jmax = 100, 420
    elif size == 200:
        imin, imax = 100, 210
        jmin, jmax = 50, 210
    else:
        imin, imax = 20, 42
        jmin, jmax = 10, 42

    selected_data = data[jmin:jmax, imin:imax, selected_channels]
    normalized = np.zeros_like(selected_data, dtype=np.float32)
    norm_params = []

    for i in range(1):  # Only normalize first channel (pressure)
        channel = selected_data[:, :, i]

        if pressure_norm_mode == "default":
            min_val, max_val = channel.min(), channel.max()
            norm_params.append(("default", min_val, max_val))
            channel_norm = (channel - min_val) / (max_val - min_val)

        elif pressure_norm_mode == "percentile":
            min_val, max_val = np.percentile(channel, 2), np.percentile(channel, 98)
            norm_params.append(("percentile", min_val, max_val))
            channel_norm = np.clip((channel - min_val) / (max_val - min_val), 0, 1)

        elif pressure_norm_mode == "zscore":
            mean, std = channel.mean(), channel.std()
            norm_params.append(("zscore", mean, std))
            channel_norm = np.clip((channel - mean) / std, -2, 2)
            channel_norm = (channel_norm + 2) / 4  # scale to [0,1]

        elif pressure_norm_mode == "log":
            channel = np.log1p(channel)
            min_val, max_val = channel.min(), channel.max()
            norm_params.append(("log", min_val, max_val))
            channel_norm = (channel - min_val) / (max_val - min_val)

        else:
            raise ValueError("Unknown pressure_norm_mode")

        normalized[:, :, i] = 255 * channel_norm

    if return_params:
        return normalized.astype(np.uint8), norm_params
    return normalized.astype(np.uint8)


def denormalize_image(normalized_image, norm_params, as_red_rgb=True):
    # Assume only one channel
    channel = normalized_image[:, :, 0].astype(np.float32) / 255.0
    method = norm_params[0][0]

    if method == "default":
        _, min_val, max_val = norm_params[0]
        denorm = channel * (max_val - min_val) + min_val

    elif method == "percentile":
        _, min_val, max_val = norm_params[0]
        denorm = channel * (max_val - min_val) + min_val

    elif method == "zscore":
        _, mean, std = norm_params[0]
        z = channel * 4 - 2  # rescale back from [0,1] to [-2,2]
        denorm = z * std + mean

    elif method == "log":
        _, min_val, max_val = norm_params[0]
        log_channel = channel * (max_val - min_val) + min_val
        denorm = np.expm1(log_channel)

    else:
        raise ValueError("Unknown normalization method")

    if as_red_rgb:
        # Normalize back to [0, 255] for visualization
        min_val, max_val = denorm.min(), denorm.max()
        red_channel = 255 * (denorm - min_val) / (max_val - min_val + 1e-8)
        red_channel = red_channel.astype(np.uint8)

        # Stack into RGB: Red only
        rgb_image = np.zeros((*red_channel.shape, 3), dtype=np.uint8)
        rgb_image[:, :, 0] = red_channel
        return rgb_image

    return denorm  # Just return grayscale values if not in RGB


modes = ["default", "percentile", "zscore", "log"]

for mode in modes:
    image, params = preprocess_image("/scratch/network/For_Nathan/input_data.npy", 1000, pressure_norm_mode=mode, return_params=True)
    restored = denormalize_image(image, params, as_red_rgb=True)

    fig, axs = plt.subplots(1, 2, figsize=(10, 5))
    
    axs[0].imshow(image, origin="lower")
    axs[0].set_title(f"Normalized ({mode})")
    axs[0].axis("off")

    axs[1].imshow(restored, origin="lower")
    axs[1].set_title(f"Denormalized Red ({mode})")
    axs[1].axis("off")

    plt.tight_layout()
    plt.show()
```
