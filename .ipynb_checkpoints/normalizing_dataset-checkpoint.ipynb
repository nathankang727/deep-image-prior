I decided to create a new notebook separate from data_loading.ipynb that focuses on normalizing Mia's input dataset so my work is more organized.

For reference, each of the 8 channels are as follows:

0. Coarse PF output
1. Hydraulic conductivity of the top soil layer
2. Manning’s roughness
3. x-slope
4. y-slope
5. Upstream area
6. Topographic wetness index
7. Elevation

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors

# Print out minimum and maximum values and plot a histogram for each channel.
data = np.load("/scratch/network/For_Nathan/input_data_100m.npy")  # shape: (H, W, C)
data[data < 0] = 0

imin, imax = 200, 420
jmin, jmax = 100, 420

selected_data = data[jmin:jmax, imin:imax, 0:8]

for i in range(8):
    channel = selected_data[:, :, i].flatten()
    min_val, max_val = channel.min(), channel.max()
    print("Channel " + str(i) + ": Min " + str(min_val) + ", Max " + str(max_val))
    plt.hist(channel)
    plt.show()
    plt.close()
```

```python
# Normalize the channels. Since each channel might have their own normalization scheme, I may consider
# implementing them individually one-by-one.

import numpy as np
import matplotlib.pyplot as plt

def preprocess_data(path, size, selected_channels=range(0, 8)):
    # Load and clip negative values
    data = np.load(path)  # shape: (H, W, C)
    data[data < 0] = 0
    
    # Crop based on resolution
    if size == 100:
        imin, imax = 200, 420
        jmin, jmax = 100, 420
    elif size == 1000:
        imin, imax = 20, 42
        jmin, jmax = 10, 42
    elif size == 500:
        imin, imax = 40, 84
        jmin, jmax = 20, 84
    elif size == 200:
        imin, imax = 100, 210
        jmin, jmax = 50, 210

    selected_data = data[jmin:jmax, imin:imax, selected_channels]
    normalized = np.zeros_like(selected_data, dtype=np.float32)

    for i in range(3):
        channel = selected_data[:, :, i]
 
#         if i < 2 or (i > 2 and i < 6):  # Log normalization for channels 0-1, 3-5

        raw = selected_data[:, :, i]
        print(f"Channel {i} stats BEFORE log: min={raw.min()}, max={raw.max()}, mean={raw.mean()}")
        
        if i < 2: # channels 0 and 5 are log normalized
            channel = np.log1p(channel)
        min_val, max_val = channel.min(), channel.max()
        
        print (channel.min(), channel.max())

        # Avoid divide-by-zero
        if max_val > min_val:
            normalized[:, :, i] = 255 * (channel - min_val) / (max_val - min_val)
        print (normalized.min(), normalized.max())
    
    return normalized.astype(np.uint8)


normalized_data = preprocess_data("/scratch/network/For_Nathan/input_data_200m.npy", 200, selected_channels=(0, 5, 7))

plt.imshow(normalized_data, origin="lower")
plt.axis("off")
# plt.show()
plt.savefig("data/sr/input_200x200_normalized.png", bbox_inches="tight")
plt.close()
# print(normalized_data)
```
